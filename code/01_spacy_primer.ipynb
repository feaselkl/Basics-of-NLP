# SpaCy Demonstration

This notebook provides a brief walkthrough of the [spaCy](https://spacy.io/) library for Natural Language Processing (NLP) in Python. We'll cover:

- Installing and loading spaCy and its English model
- Part-of-speech (POS) tagging
- Named Entity Recognition (NER)
- Dependency parsing
- Extracting noun chunks and verb phrases
- Visualizing the parse tree

## 1. Install and Load spaCy and the English Model

```python
# If running for the first time, uncomment the following lines to install the model:
#!python -m spacy download en_core_web_lg
```

```python
import spacy
nlp = spacy.load('en_core_web_lg')
```

## 2. Part-of-Speech Tagging

Let's analyze a sample sentence and display the part-of-speech tags for each token.

```python
sentence = "On one otherwise normal Tuesday evening I had the chance to live the American dream. I was able to throw my incompetent jackass of a boss from a fourteenth-story window."
doc = nlp(sentence)
for token in doc:
    print(f"{token.text:15} {token.pos_:10} {token.tag_:6} {spacy.explain(token.tag_)}")
```

## 3. Named Entity Recognition (NER)

Let's use a longer sentence and extract named entities.

```python
long_sentence = (
    """The Battle of Liegnitz on 15 August 1760 saw Frederick the Great's Prussian Army defeat the Austrian army under Ernst von Laudon during the Third Silesian War.

    The armies collided around the town of Liegnitz (now Legnica, Poland) in Lower Silesia. 
    Laudon's Austrian cavalry attacked the Prussian position in the early morning but were beaten back by General Zieten's Hussars. 
    An artillery duel emerged which was eventually won for the Prussians when a shell hit an Austrian powder wagon. 
    The Austrian infantry then proceeded to attack the Prussian line, but was met with concentrated artillery fire. 
    A Prussian infantry counter-attack led by the Regiment Anhalt-Bernburg on the left forced the Austrians into retreat. 
    Notably, the Anhalt-Bernburgers charged Austrian cavalry with bayonets, a rare example of infantry assaulting cavalry.

    Shortly after dawn the major action was over but Prussian artillery fire continued to harass the Austrians. 
    General Leopold von Daun arrived and, learning of Laudon's defeat, decided not to attack despite his soldiers being fresh.
    """
)
doc2 = nlp(long_sentence)
for ent in doc2.ents:
    print(f"{ent.text:30} {ent.label_:10} {spacy.explain(ent.label_)}")
```

## 4. Dependency Parsing

Let's analyze the syntactic dependencies in a sentence. The attributes we will look at are:

* `text` - The original text of a token
* `pos_` - A string value representing the token's part of speech
* `head` - The original text of the token's head
* `head.pos` - A string value representing the token head's part of speech
* `dep_` - A string value of the arc label (describing the syntactic relation that connects the child to the head)
* `explain()` - Translate an arc label or other SpaCy descriptor into a full string


```python
s = "By this point Frederick was increasingly concerned by the Russian advance from the east and marched to counter it."
doc3 = nlp(s)
print(f"{"TOKEN":12} {"POS":6} {"HEAD":12} {"HD_POS":6} {"DEP":10} {"EXPLANATION"}")
for token in doc3:
    print(f"{token.text:12} {token.pos_:6} {token.head.text:12} {token.head.pos_:6} {token.dep_:10} {spacy.explain(token.dep_)}")
```

## 5. Noun Chunks and Verb Phrases

Let's extract noun chunks and verb phrases from the sentence.

```python
print("Noun Chunks:")
for chunk in doc2.noun_chunks:
    print(f"- {chunk.text}")

# Extracting verb phrases is not directly supported, but we can get verbs and their children
print("\nVerb Phrases:")
for token in doc2:
    if token.pos_ == "VERB":
        phrase = ' '.join([child.text for child in token.subtree])
        print(f"- {phrase}")
```

## 6. Visualizing the Parse Tree

Parse trees are the computer science version of the sentence diagramming we looked at early on.

spaCy provides built-in visualization tools via `spacy.displacy`. In Jupyter, you can render them directly.

```python
from spacy import displacy
displacy.render(doc, style="dep", jupyter=True)
```
