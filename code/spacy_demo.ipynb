# SpaCy Demonstration

This notebook provides a brief walkthrough of the [spaCy](https://spacy.io/) library for Natural Language Processing (NLP) in Python. We'll cover:

- Installing and loading spaCy and its English model
- Part-of-speech (POS) tagging
- Named Entity Recognition (NER)
- Dependency parsing
- Extracting noun chunks and verb phrases
- Visualizing the parse tree

## 1. Install and Load spaCy and the English Model

```python
# If running for the first time, uncomment the following lines to install the model:
# !python -m spacy download en_core_web_sm
import spacy
nlp = spacy.load('en_core_web_sm')
```

## 2. Part-of-Speech Tagging

Let's analyze a sample sentence and display the part-of-speech tags for each token.

```python
sentence = "SpaCy is a popular library for advanced Natural Language Processing in Python."
doc = nlp(sentence)
for token in doc:
    print(f"{token.text:15} {token.pos_:10} {token.tag_:6} {spacy.explain(token.tag_)}")
```

## 3. Named Entity Recognition (NER)

Let's use a longer sentence and extract named entities.

```python
long_sentence = (
    "Apple is looking at buying U.K. startup for $1 billion in 2025, "
    "according to sources familiar with the matter."
)
doc2 = nlp(long_sentence)
for ent in doc2.ents:
    print(f"{ent.text:30} {ent.label_:10} {spacy.explain(ent.label_)}")
```

## 4. Dependency Parsing

Let's analyze the syntactic dependencies in a sentence.

```python
for token in doc2:
    print(f"{token.text:12} {token.dep_:10} {token.head.text:12} {token.pos_:6}")
```

## 5. Noun Chunks and Verb Phrases

Let's extract noun chunks and verb phrases from the sentence.

```python
print("Noun Chunks:")
for chunk in doc2.noun_chunks:
    print(f"- {chunk.text}")

# Extracting verb phrases is not directly supported, but we can get verbs and their children
print("\nVerb Phrases:")
for token in doc2:
    if token.pos_ == "VERB":
        phrase = ' '.join([child.text for child in token.subtree])
        print(f"- {phrase}")
```

## 6. Visualizing the Parse Tree

spaCy provides built-in visualization tools via `spacy.displacy`. In Jupyter, you can render them directly.

```python
from spacy import displacy
displacy.render(doc2, style="dep", jupyter=True)
```
